```python
df = pd.read_csv("AirQuality_300_final_no_city.csv")

# Convert Date
df['Date'] = pd.to_datetime(
    df['Date'], format='mixed', errors='coerce'
)
```

```python
sns.set_theme(style="whitegrid")
```

```python
# 1. Line Plot
plt.figure(figsize=(10, 5))
data = df.dropna(subset=['Date', 'CO_GT']).sort_values('Date')

sns.lineplot(data=data, x='Date', y='CO_GT', color='blue')

plt.title("CO Concentration Over Date")
plt.xlabel("Date")
plt.ylabel("CO_GT")
plt.xticks(rotation=45)
plt.show()
```

```python
# 2. Count Plot
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x='CO_level',
    color='skyblue'
)

plt.title("CO Level Count")
plt.xlabel("CO Level")
plt.ylabel("Count")
plt.show()
```

```python
# 3. Histogram
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x='CO_GT',
    bins=10,
    kde=True,
    color='green'
)

plt.title("CO Distribution")
plt.xlabel("CO_GT")
plt.ylabel("Frequency")
plt.show()
```

```python
# 4. Scatter Plot
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='T',
    y='CO_GT',
    hue='CO_level'
)

plt.title("Temperature vs CO")
plt.xlabel("Temperature")
plt.ylabel("CO_GT")
plt.show()
```

```python
# 8. Heatmap
plt.figure(figsize=(12, 8))

corr = df.select_dtypes(include='number').corr()

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    linewidths=0.5
)

plt.title("Air Quality Correlation Heatmap")
plt.show()
```
